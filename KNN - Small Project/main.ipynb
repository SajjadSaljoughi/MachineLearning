KNN Visualization
In this notebook, we will implement a simple K-Nearest Neighbors (KNN) algorithm from scratch and visualize how it works using a simple dataset of clothes. We will use two features: width and height, to classify clothes as either shirts or pants or hats.

Now We Should Create Size of Clothes Dataset :
- shirt width ≈ 50cm
- shirt height ≈ 70cm
- pants width ≈ 30cm
- pants height ≈ 100cm
- hat width ≈ 25cm
- hat height ≈ 20cm

and in Y_Train, 0 represents shirts, 1 represents pants and 2 represents hats.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 200
shirt_width = np.random.normal(50, 15, n) + 25
shirt_height = np.random.normal(70, 15, n) + 25
pants_width = np.random.normal(30, 15, n) + 25
pants_height = np.random.normal(100, 15, n) + 25
hat_width = np.random.normal(25, 15, n) + 25
hat_height = np.random.normal(20, 15, n) + 25
```

```python
plt.scatter(shirt_width, shirt_height, color='red', label='Shirts')
plt.scatter(pants_width, pants_height, color='blue', label='Pants')
plt.scatter(hat_width, hat_height, color='green', label='Hats')
plt.xlabel('Width')
plt.ylabel('Height')
plt.title('Clothing Items')
plt.legend()
plt.show()
```

```python
width = np.concatenate((shirt_width, pants_width, hat_width))
height = np.concatenate((shirt_height, pants_height, hat_height))
X_train = np.array([width, height]).T
```

```python
Y_train = np.array([0]*n + [1]*n + [2]*n)
```

```python
def euclidean_distance(x1, x2):
    return np.sqrt(np.sum((x1 - x2) ** 2))
```

```python
new_clothing_item = np.array([60, 110])
distances = []
for x in X_train:
    d = euclidean_distance(x, new_clothing_item)
    distances.append(d)
k = 5
nearest_indices = np.argsort(distances)[:k]
nearest_labels = Y_train[nearest_indices]
predicted_label = np.bincount(nearest_labels).argmax()
if predicted_label == 0:
    print("The new clothing item is predicted to be a Shirt.")
elif predicted_label == 1:
    print("The new clothing item is predicted to be Pants.")
else:
    print("The new clothing item is predicted to be a Hat.")
```
