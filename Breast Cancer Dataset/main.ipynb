```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
from sklearn.neighbors import KNeighborsClassifier
from knn import KNN
```

```python
(X,Y) = load_breast_cancer(return_X_y=True)
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2)

knn_3 = KNN(3)
knn_3.fit(X_train,Y_train)
accuracy_k3 = knn_3.evaluate(X_test,Y_test)

knn_3_sklearn = KNeighborsClassifier(n_neighbors=3)
knn_3_sklearn.fit(X_train,Y_train)
accuracy_k3_sklearn = knn_3_sklearn.score(X_test,Y_test)

knn_5 = KNN(5)
knn_5.fit(X_train,Y_train)
accuracy_k5 = knn_5.evaluate(X_test,Y_test)

knn_5_sklearn = KNeighborsClassifier(n_neighbors=5)
knn_5_sklearn.fit(X_train,Y_train)
accuracy_k5_sklearn = knn_5_sklearn.score(X_test,Y_test)

knn_7 = KNN(7)
knn_7.fit(X_train,Y_train)
accuracy_k7 = knn_7.evaluate(X_test,Y_test)

knn_7_sklearn = KNeighborsClassifier(n_neighbors=7)
knn_7_sklearn.fit(X_train,Y_train)
accuracy_k7_sklearn = knn_7_sklearn.score(X_test,Y_test)

knn_9 = KNN(9)
knn_9.fit(X_train,Y_train)
accuracy_k9 = knn_9.evaluate(X_test,Y_test)

knn_9_sklearn = KNeighborsClassifier(n_neighbors=9)
knn_9_sklearn.fit(X_train,Y_train)
accuracy_k9_sklearn = knn_9_sklearn.score(X_test,Y_test)

knn_11 = KNN(11)
knn_11.fit(X_train,Y_train)
accuracy_k11 = knn_11.evaluate(X_test,Y_test)

knn_11_sklearn = KNeighborsClassifier(n_neighbors=11)
knn_11_sklearn.fit(X_train,Y_train)
accuracy_k11_sklearn = knn_11_sklearn.score(X_test,Y_test)
```

```python
knn_list = [3,5,7,9,11]
accuracy_list = [accuracy_k3,accuracy_k5,accuracy_k7,accuracy_k9,accuracy_k11]
accuracy_list_sklearn = [accuracy_k3_sklearn,accuracy_k5_sklearn,accuracy_k7_sklearn,accuracy_k9_sklearn,accuracy_k11_sklearn]
plt.plot(knn_list,accuracy_list)
plt.plot(knn_list,accuracy_list_sklearn)
plt.show()
```

```python
y_pred = knn_5.predict(X_test)
confusion_matrix = np.zeros((2,2))
for i in range(len(y_pred)):
    if y_pred[i] == 0 and Y_test[i] == 0:
        confusion_matrix[0,0] += 1
    elif y_pred[i] == 0 and Y_test[i] == 1:
        confusion_matrix[0,1] += 1
    elif y_pred[i] == 1 and Y_test[i] == 0:
        confusion_matrix[1,0] += 1
    elif y_pred[i] == 1 and Y_test[i] == 1:
        confusion_matrix[1,1] += 1
plt.xlabel('Predicted label')
plt.ylabel('True label')
plt.title('Confusion Matrix')
for i in range(confusion_matrix.shape[0]):
    for j in range(confusion_matrix.shape[1]):
        plt.text(j, i, confusion_matrix[i, j],
                 ha="center", va="center", color="black")
plt.imshow(confusion_matrix,cmap="Reds")
plt.colorbar()
plt.show()
```
