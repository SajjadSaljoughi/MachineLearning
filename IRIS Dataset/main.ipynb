```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
from knn import KNN
```

```python
iris = load_iris()
X = iris.data
Y = iris.target
X_train,X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2)
knn_3 = KNN(3)
knn_3.fit(X,Y)
accuracy_k3 = knn_3.evaluate(X_test,Y_test)
knn_5 = KNN(5)
knn_5.fit(X,Y)
accuracy_k5 = knn_5.evaluate(X_test,Y_test)
knn_7 = KNN(7)
knn_7.fit(X,Y)
accuracy_k7 = knn_7.evaluate(X_test,Y_test)
knn_9 = KNN(9)
knn_9.fit(X,Y)
accuracy_k9 = knn_9.evaluate(X_test,Y_test)
knn_11 = KNN(7)
knn_11.fit(X,Y)
accuracy_k11 = knn_11.evaluate(X_test,Y_test)
```

```python
knn_list = [3,5,7,9,11]
accuracy_list = [accuracy_k3,accuracy_k5,accuracy_k7,accuracy_k9,accuracy_k11]
plt.plot(knn_list,accuracy_list)
plt.xlabel('K Values')
plt.ylabel('Accuracy')
plt.savefig("output/Accuracy.png")
plt.show()
```

```python
y_pred = knn_3.predict(X_test)
confusion_matrix = np.zeros((3,3))
for i in range(len(y_pred)):
    if y_pred[i] == 0 and Y_test[i] == 0:
        confusion_matrix[0,0] += 1
    elif y_pred[i] == 0 and Y_test[i] == 1:
        confusion_matrix[0,1] += 1
    elif y_pred[i] == 0 and Y_test[i] == 2:
        confusion_matrix[0,2] += 1
    elif y_pred[i] == 1 and Y_test[i] == 0:
        confusion_matrix[1,0] += 1
    elif y_pred[i] == 1 and Y_test[i] == 1:
        confusion_matrix[1,1] += 1
    elif y_pred[i] == 1 and Y_test[i] == 2:
        confusion_matrix[1,2] += 1
    elif y_pred[i] == 2 and Y_test[i] == 0:
        confusion_matrix[2,0] += 1
    elif y_pred[i] == 2 and Y_test[i] == 1:
        confusion_matrix[2,1] += 1
    elif y_pred[i] == 2 and Y_test[i] == 2:
        confusion_matrix[2,2] += 1
plt.xlabel('Predicted label')
plt.ylabel('True label')
plt.title('Confusion Matrix')
for i in range(confusion_matrix.shape[0]):
    for j in range(confusion_matrix.shape[1]):
        plt.text(j, i, confusion_matrix[i, j],
                 ha="center", va="center", color="black")
plt.imshow(confusion_matrix,cmap="Reds")
plt.colorbar()
plt.savefig("output/ConfusionMatrix.png")
plt.show()
```
