```python
import matplotlib.pyplot as plt
import cv2
from FindNemo import FindNemo
```

```python
for i in range(1,7):
    train_image = cv2.imread(f"images\\nemo_1.jpg")
    find_nemo = FindNemo(train_image)
    new_nemo = cv2.imread(f"images\\nemo_{i}.jpg")
    result = find_nemo.remove_background(new_nemo)
    result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)
    cv2.imwrite(f'output\\nemo_{i}.jpg', result_rgb)
```
