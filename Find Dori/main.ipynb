```python
import matplotlib.pyplot as plt
import cv2
from FindDori import FindDori
```

```python
for i in range(1,5):
    train_image = cv2.imread(f"images\\Dori1.jpg")
    find_dori = FindDori(train_image)
    new_dori = cv2.imread(f"images\\Dori{i}.jpg")
    result = find_dori.remove_background(new_dori)
    result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)
    cv2.imwrite(f'output\\Dori{i}.jpg', result_rgb)
```
