```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.colors import ListedColormap
```

```python
female_data = pd.read_csv('ANSUR II FEMALE Public.csv', encoding='latin-1')
male_data = pd.read_csv('ANSUR II MALE Public.csv', encoding='latin-1')
```

```python
data = pd.concat([female_data, male_data])
```

```python
data.head()
```

```python
data["weightkg"] = data["weightkg"] / 10 #Convert to kg
data["stature"] = data["stature"] / 10 #Convert to cm
data["Gender"] = data["Gender"].replace(["Female", "Male"], [0, 1])
```

```python
ax = plt.gca()
ax.set_facecolor('white')
cmap = ["red","blue"]
plt.scatter(data["weightkg"], data["stature"],
            c=data["Gender"],cmap=ListedColormap(cmap),
            alpha=0.1)
plt.show()
```
