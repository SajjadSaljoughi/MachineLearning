```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from knn import KNN
from sklearn.neighbors import KNeighborsClassifier
```

```python
nemo_1 = cv2.imread("input//nemo_1.jpg")
nemo_1 = cv2.resize(nemo_1, (0, 0), fx=0.25, fy=0.25)
nemo_1 = cv2.cvtColor(nemo_1, cv2.COLOR_BGR2RGB)
plt.imshow(nemo_1)
plt.show()
```

```python
pixels_list_rgb = nemo_1.reshape(-1, 3)
pixels_list_rgb.shape
```

```python
fig = plt.figure(figsize=(4, 4))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(pixels_list_rgb[:, 0], pixels_list_rgb[:, 1], pixels_list_rgb[:, 2],
           c=pixels_list_rgb / 255)
plt.show()
```

```python
nemo_1_hsv = cv2.cvtColor(nemo_1, cv2.COLOR_RGB2HSV)
pixels_list_hsv = nemo_1_hsv.reshape(-1, 3)
fig = plt.figure(figsize=(4, 4))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(pixels_list_hsv[:, 0], pixels_list_hsv[:, 1], pixels_list_hsv[:, 2],
           c=pixels_list_rgb / 255)
plt.show()
```

```python
plt.imshow(nemo_1_hsv)
plt.show()
```

```python
light_orange = (1,190,200)
dark_orange = (18,255,255)
mask_orange = cv2.inRange(nemo_1_hsv,
                   light_orange,
                   dark_orange)
plt.imshow(mask_orange, cmap='gray')
plt.show()
```

```python
light_white = (0,0,200)
dark_white = (145,60,255)
mask_white = cv2.inRange(nemo_1_hsv,
                   light_white,
                   dark_white)
plt.imshow(mask_white, cmap='gray')
plt.show()
```

```python
final_mask = mask_orange + mask_white
plt.imshow(final_mask, cmap='gray')
plt.show()
```

```python
final_nemo_1 = cv2.bitwise_and(nemo_1, nemo_1, mask=final_mask)
plt.imshow(final_nemo_1)
plt.show()
```

```python
X_train = pixels_list_hsv / 255
X_train.shape
```

```python
Y_train = final_mask.reshape(-1,) // 255
Y_train.shape
```

```python
knn = KNN(k=3)
knn.fit(X_train, Y_train)
# knn = KNeighborsClassifier(n_neighbors=3)
# knn.fit(X_train, Y_train)
```

```python
new_pixel = np.array([10,200,255]) / 255
knn.predict([new_pixel])
```

```python
nemo_2 = cv2.imread("input//nemo_2.jpg")
nemo_2 = cv2.resize(nemo_2, (0, 0), fx=0.25, fy=0.25)
nemo_2 = cv2.cvtColor(nemo_2, cv2.COLOR_BGR2RGB)
nemo_2_hsv = cv2.cvtColor(nemo_2, cv2.COLOR_RGB2HSV)
X_test = nemo_2_hsv.reshape(-1,3) / 255
X_test.shape
```

```python
Y_predict = knn.predict(X_test)
```

```python
output = Y_predict.reshape(nemo_2_hsv.shape[:2])
plt.imshow(output, cmap='gray')
plt.show()
```
