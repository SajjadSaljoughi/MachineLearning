```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
n = 100
apple_width = np.random.normal(6,1,100)
apple_height = np.random.normal(6,1,100)
banana_width = np.random.normal(4,1,100)
banana_height = np.random.normal(8,1,100)
```

```python
x_width = 3
x_height = 9
```

```python
ax = plt.gca()
ax.set_facecolor('white')
plt.scatter(apple_width,apple_height,
            color='red',marker='.')
plt.scatter(banana_width,banana_height,
            color='#E5E500',marker='.')
plt.scatter(x_width,x_height,
            color='blue',marker='o')
plt.title('Apple and Banana')
plt.xlabel('Width')
plt.ylabel('Height')
plt.show()
```

```python
#KNN K Nearest Neighbor
width = np.concatenate((apple_width,banana_width))
length = np.concatenate((apple_height,banana_height))
X_train = np.array([width,length]).T
Y_apples = np.zeros(n, dtype = int)
Y_banana = np.ones(n, dtype = int)
Y_train = np.concatenate((Y_apples,Y_banana))
#This is a dataset
```

```python
def euclidean_distance(x1, x2):
    return np.sqrt(np.sum((x1-x2) ** 2))
```

```python
new_fruit = np.array([3,8])
distances = []
for x in X_train:
    distances.append(euclidean_distance(new_fruit,x))
```

```python
k = 5
nearest_neighbour = np.argsort(distances)[0:k]
nearest_neighbour
```

```python
result = np.argmax(np.bincount(Y_train[nearest_neighbour]))
if result == 0:
    print("🍎")
elif result == 1:
    print("🍌")
```
