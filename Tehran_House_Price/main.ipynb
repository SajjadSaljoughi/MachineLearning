```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from LLS import LLS
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import RidgeCV
from sklearn.metrics import mean_squared_error,mean_absolute_error,root_mean_squared_error
```

```python
data = pd.read_csv("housePrice.csv")
data.head()
```

```python
#Update Price for 24/09/2026
#Dollar in Iran = 230,000
data["Price"] = data["Price(USD)"].astype(np.float64) * 230
data.head()
```

```python
address_data = data.nlargest(5, "Price")
address_data
```

```python
#Pre-Process
data["Parking"] = data["Parking"].replace([False, True], [0, 1])
data["Warehouse"] = data["Warehouse"].replace([False, True], [0, 1])
data["Elevator"] = data["Elevator"].replace([False, True], [0, 1])
```

```python
data.head()
```

```python
X = data[["Area", "Room", "Parking", "Warehouse", "Elevator"]].to_numpy(dtype=np.float64)
Y = np.array(data["Price"])
```

```python
X.shape, Y.shape
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2)
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape
```

```python
Y_train = Y_train.reshape(-1, 1)
X_train = np.array(X_train)
X_train.dtype
```

```python
lls = LLS()
lls.fit(X_train, Y_train)
```

```python
loss_mae = lls.evaluate(X_test, Y_test,metric="mae")
loss_mse = lls.evaluate(X_test, Y_test,metric="mse")
loss_rmse = lls.evaluate(X_test, Y_test,metric="rmse")
```

```python
print(f"loss mae = {loss_mae}")
print(f"loss mse = {loss_mse}")
print(f"loss rmse = {loss_rmse}")
```

```python
#LinearRegression
lr = LinearRegression()
lr.fit(X_train, Y_train)
y_pred_lr = lr.predict(X_test)

loss_lr_mae = mean_absolute_error(Y_test, y_pred_lr)
loss_lr_mse = mean_squared_error(Y_test, y_pred_lr)
loss_lr_rmse = root_mean_squared_error(Y_test, y_pred_lr)
```

```python
print(f"loss mae = {loss_lr_mae}")
print(f"loss mse = {loss_lr_mse}")
print(f"loss rmse = {loss_lr_rmse}")
```

```python
#RidgeCV
rcv = RidgeCV()
rcv.fit(X_train, Y_train)
y_pred_rcv = rcv.predict(X_test)

loss_rcv_mae = mean_absolute_error(Y_test, y_pred_rcv)
loss_rcv_mse = mean_squared_error(Y_test, y_pred_rcv)
loss_rcv_rmse = root_mean_squared_error(Y_test, y_pred_rcv)
```

```python
print(f"loss mae = {loss_rcv_mae}")
print(f"loss mse = {loss_rcv_mse}")
print(f"loss rmse = {loss_rcv_rmse}")
```
