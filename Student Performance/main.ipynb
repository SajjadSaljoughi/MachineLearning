```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from LLS import LLS
```

```python
X_train = np.array([
    1, 1.5, 3, 1.5, 2, 2.5, 4, 3.5, 1, 2,
    2.5, 1.5, 1, 4, 2.5, 2, 2, 1, 1, 1.5,
    3, 2, 1.5, 3.5, 2.5, 1, 3, 2, 4, 1.5,
    2, 2.5, 3, 1, 3.5, 2, 1.5, 2.5, 4, 1,
    2, 3, 1.5, 2.5, 3.5, 1, 2, 2.5, 3, 1.5,
    4, 2, 1, 3, 2.5, 1.5, 3.5, 2, 2.5, 1,
    3, 4, 1.5, 2, 3, 2.5, 1, 3.5, 2, 1.5,
    2.5, 3, 4, 1, 2, 1.5, 3.5, 2.5, 3, 1,
    2, 3, 1.5, 4, 2.5, 1, 3, 2, 3.5, 1.5,
    2, 2.5, 4, 1, 3, 1.5, 2, 3.5, 2.5, 1
])
Y_train = np.array([
    5, 8, 15, 7, 10, 13, 20, 18, 6, 11,
    12, 9, 4, 19, 14, 10, 12, 5, 7, 8,
    15, 11, 8, 17, 13, 6, 14, 10, 20, 7,
    11, 13, 16, 5, 18, 10, 8, 14, 19, 6,
    11, 15, 9, 13, 17, 5, 10, 14, 16, 8,
    20, 9, 5, 15, 13, 9, 18, 11, 12, 6,
    14, 20, 8, 10, 15, 13, 5, 17, 11, 8,
    14, 16, 19, 5, 11, 7, 18, 13, 15, 6,
    10, 14, 8, 20, 12, 5, 16, 9, 18, 8,
    10, 13, 19, 5, 15, 7, 11, 17, 14, 6
])
```

```python
plt.scatter(X_train,Y_train)
plt.show()
```

```python
lls = LLS()
lls.fit(X_train,Y_train)
```

```python
y_pred = lls.predict(1.5)
y_pred
```

```python
a = lls.calculate_slope()
a.shape
```

```python
plt.scatter(X_train,Y_train)
plt.plot(X_train,0.9 * X_train,color='red')
plt.plot(X_train,a * X_train,color='blue')
plt.show()
```

```python
#scipy
slope, intercept, r_value, p_value, std_err = stats.linregress(X_train, Y_train)
plt.scatter(X_train,Y_train)
plt.plot(X_train,0.9 * X_train,color='red',label='Random Linear Regression')
plt.plot(X_train,a * X_train,color='blue',label='My Linear Regression')
plt.plot(X_train,slope * X_train + intercept,color='green',label='SciPy Linear Regression')
plt.xlabel('hour')
plt.ylabel('Score')
plt.legend(loc='upper left')
plt.savefig("output/lls.png")
plt.show()
```

```python
new_hour = 2.5
my_pred = lls.predict(new_hour)
scipy_pred = new_hour * slope + intercept
print(my_pred)
print(scipy_pred)
```
