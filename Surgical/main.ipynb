```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from tqdm import tqdm
from perceptron import Perceptron
```

```python
df = pd.read_csv("Surgical-deepnet.csv")
df.head()
```

```python
X = df.drop("complication",axis=1).copy().values
Y = df["complication"].copy().values

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2)
X_train.shape, Y_train.shape, X_test.shape, Y_test.shape
```

```python
model1 = Perceptron(learning_rate=0.001,input_length=X_train.shape[1],
                    epochs=256)
model1.fit(X_train, Y_train,
           X_test, Y_test,
           "sigmoid")
```

```python
plt.plot(model1.loss_history,'b')
plt.plot(model1.test_loss_history,'r')
plt.title('Model Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend(['Train', 'Test'], loc='upper right')
plt.show()
```

```python
plt.plot(model1.accuracy_history,'b')
plt.plot(model1.test_accuracy_history,'r')
plt.title('Model Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend(['Train', 'Test'], loc='upper left')
plt.show()
```
