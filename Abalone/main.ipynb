```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from perceptron import Perceptron
```

```python
data = pd.read_csv('abalone.csv')
data.head()
```

```python
plt.scatter(data["Height"], data["Length"])
plt.show()
```

we can find 2 data column for regression


```python
data["Sex"] = data["Sex"].replace(["M", "F", "I"], [0, 1, 2])
```

```python
data.corr()
```

```python
plt.scatter(data["Length"], data["Height"])
plt.show()
```

```python
X = np.array(data["Length"])
Y = np.array(data["Height"])
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2)
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape
```

```python
X_train = X_train.reshape(-1, 1)
X_test = X_test.reshape(-1, 1)
Y_train = Y_train.reshape(-1, 1)
Y_test = Y_test.reshape(-1, 1)
```

```python
perceptron = Perceptron(1, 0.001, 0.1, 256)
perceptron.fit(X_train, Y_train)
```

```python
Y_pred = perceptron.predict(X_test)
perceptron.show_plot(X_train, Y_train, Y_pred)
```
