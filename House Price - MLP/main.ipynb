```python
import numpy as np
import tensorflow as tf
import pandas as pd
```

```python
model = tf.keras.models.load_model("house_price.h5")
```

```python
my_house = np.array([[
    7,     # OverallQual
    1500,  # GrLivArea
    2,     # GarageCars
    800,   # TotalBsmtSF
    2005,  # YearBuilt
    2,     # FullBath
    6,     # TotRmsAbvGrd
    100    # WoodDeckSF
]])
model.predict(my_house)
```

```python
test_data = pd.read_csv("test.csv")
test_data.head()
```

```python
X_test = test_data[[
    "OverallQual",
    "GrLivArea",
    "GarageCars",
    "TotalBsmtSF",
    "YearBuilt",
    "FullBath",
    "TotRmsAbvGrd",
    "WoodDeckSF"
]].values
```

```python
Y_pred = model.predict(X_test)
```

```python
pred_csv_data = {
    "Id":[],
    "SalePrice":[]
}
for i in range(len(Y_pred)):
    pi = test_data["Id"].iloc[i]
    pred_csv_data["Id"].append(pi)
    pred_csv_data["SalePrice"].append(Y_pred[i])
df_pred_csv = pd.DataFrame(pred_csv_data)
df_pred_csv.to_csv("SS_Predict_1.csv", index=False)
```
